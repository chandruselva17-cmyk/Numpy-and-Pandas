```python
import numpy as np
import pandas as pd
```

```python
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

print("--- 1D Array Properties ---")
print("Array:", temperatures_w1)
print("Shape:", temperatures_w1.shape)
print("Data Type:", temperatures_w1.dtype)
print("Number of Elements:", temperatures_w1.size)
```

```python
temperatures_w1_f = (temperatures_w1 * 9/5) + 32

max_temp = temperatures_w1.max()
min_temp = temperatures_w1.min()
mean_temp = temperatures_w1.mean()

print("\n--- Array Operations ---")
print("Fahrenheit:", temperatures_w1_f)
print(f"Max Temp: {max_temp}°C")
print(f"Min Temp: {min_temp}°C")
print(f"Mean Temp: {mean_temp:.2f}°C")
```

```python
first_3_days = temperatures_w1[:3]

weekend = temperatures_w1[-2:]

middle_3_days = temperatures_w1[2:5]

print("\n--- 1D Slicing & Indexing ---")
print("First 3 days:", first_3_days)
print("Weekend (last 2 days):", weekend)
print("Middle 3 days:", middle_3_days)
```

```python
temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],  # Week 1
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]   # Week 2
])

print("\n--- 2D Array Properties ---")
print("Shape:", temperatures.shape)
print("Data Type:", temperatures.dtype)
print("Total Elements:", temperatures.size)

week_1_all = temperatures[0, :]
week_2_all = temperatures[1, :]

weekends_both = temperatures[:, -2:]

print("\n--- 2D Slicing ---")
print("Week 1 Temperatures:", week_1_all)
print("Week 2 Temperatures:", week_2_all)
print("Weekends (Both Weeks):\n", weekends_both)
```

```python
marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)

print("--- Initial Pandas Series ---")
print(marks)
```

```python
mark_1st = marks.iloc[0]

top_3_loc = marks.loc[['Rank1', 'Rank2', 'Rank3']]

mark_3rd_iloc = marks.iloc[2]

above_90 = marks[marks > 90]

print("\n--- Series Indexing & Slicing ---")
print("1st Rank Mark (integer position):", mark_1st)
print("Top 3 Ranks (.loc):\n", top_3_loc)
print("3rd Rank Mark (.iloc):", mark_3rd_iloc)
print("Ranks with Marks > 90:\n", above_90)
```

```python
marks['Rank1'] = 100

marks = marks.drop('Rank5')

cgpa = marks / 10

print("\n--- Manipulated Series ---")
print("Updated Marks:\n", marks)
print("\nCGPA:\n", cgpa)
```

```python
data = {
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': ['Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing',
                        'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'],
    'Region': ['North', 'South', 'North', 'East', 'West', 'North', 'East', 'West', 'South', 'North'],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
}

transactions = pd.DataFrame(data)
```

```python
print("--- Basic Information ---")
print("\nHead:\n", transactions.head())
print("\nTail:\n", transactions.tail())
print("\nShape:", transactions.shape)
print("\nColumn Names:", transactions.columns.tolist())
print("\nData Types:\n", transactions.dtypes)

print("\n--- Selected Columns ---")
print(transactions[['ProductCategory', 'Amount']])

print("\n--- Last 3 Columns ---")
print(transactions.iloc[:, -3:])

filtered_tx = transactions[(transactions['Region'] == 'North') & (transactions['Amount'] > 200)]
print("\n--- Filtered Rows (North & Amount > 200) ---")
print(filtered_tx)

print("\n--- ProductCategory Value Counts ---")
print(transactions['ProductCategory'].value_counts())

print("\n--- Unique Regions ---")
print(transactions['Region'].unique())

print("\n--- Mean Amount by Region ---")
print(transactions.groupby('Region')['Amount'].mean())
```

```python
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165

transactions['Discount'] = transactions['Amount'] * 0.10

transactions = transactions[transactions['TransactionID'] != 109]

transactions = transactions.drop(columns=['Discount'])

print("\n--- Final DataFrame ---")
print(transactions)
```
